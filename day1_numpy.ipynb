```python
import numpy as np    # "np" is just the standard nickname

speeds_list = [0, 12, 25, 41]
print(speeds_list * 2)
# [0, 12, 25, 41, 0, 12, 25, 41]   <- list repeated, NOT doubled!

speeds = np.array([0, 12, 25, 41])
print(speeds * 2)
# [ 0 24 50 82]                    <- every number doubled
```

```python
speed = np.array([0, 12, 25, 41, 55, 58, 57, 40, 22, 0])

# Columns: time (s), speed (km/h), steering angle (deg)
log = np.array([[0,  0,  0.0],
                [1, 12,  1.5],
                [2, 25,  3.0],
                [3, 41, -2.0]])          # shape (4, 3)

print(log[1, 2])       # 1.5          row 1, column 2
print(log[1])          # [ 1.  12.   1.5]   the whole row 1 (one sample)
print(log[-1])         # [ 3.  41.  -2. ]   the last row
print(log[:, 1])       # [ 0. 12. 25. 41.]  ALL rows, column 1 (the speed column)
print(log[1:3, 1:])    # rows 1-2, columns 1 to the end
                       # [[12.   1.5]
                       #  [25.   3. ]]

fast_rows = log[log[:, 1] > 20]
print(fast_rows)
# [[ 2.  25.   3.]
#  [ 3.  41.  -2.]]
```
